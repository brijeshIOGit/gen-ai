Temperature (Creativity Control)
Temperature controls how random or creative the AI's response will be. It is usually a number between 0.0 and 2.0.

Simple Analogy:

Low Temperature (0.0): Think of a serious math teacher. They give the exact same, strict answer every single time.

High Temperature (1.0 or higher): Think of a creative fiction writer. They give different, imaginative, and surprising answers every time.

When to use which:

0.0 to 0.2: For coding, math, facts, or data extraction where you need precision and zero randomness.

0.7 to 1.0: For general chat, blogging, or brainstorming where you want natural-sounding variety.

```python
response = client.chat.completions.create(
    model="gpt-4o-mini",
    temperature=0.0,
    messages=[{"role": "user", "content": "Write a 3-word slogan for coffee."}]
)

print(response.choices[0].message.content)
```

```python
encoding = tiktoken.encoding_for_model("gpt-4o-mini")
text = 'Hello World!'
tokens = encoding.encode(text)
print(f"Number of tokens in '{text}': {len(tokens)}")
```

```python
original_text = encoding.decode(tokens)
print(f"Decoded text: {original_text}")
```

```python


resp_dev_user = client.responses.create(
    model="gpt-4o-mini",
    # HIGHEST PRIORITY: Hard constraint to return ONLY code, NO explanations
    instructions="You are a strict code execution engine. Return ONLY executable Python code with ZERO text or markdown.",
    input=[
        # LOWER PRIORITY: Tries to force the model to explain in text
        {"role": "developer", "content": "Explain the concept in 5 paragraphs using simple text."},
        {"role": "user", "content": "What is a lambda function?"}
    ]
)

print(resp_dev_user.output_text)
```
